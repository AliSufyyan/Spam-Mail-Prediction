```python
import numpy as np 
import pandas as pd  
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
raw_mail_data=pd.read_csv("mail_data.csv")
```

```python
raw_mail_data.head()
```

```python
raw_mail_data.isnull().sum()
```

```python
raw_mail_data.shape
```

```python
raw_mail_data['Category'].value_counts()
```

```python
mail_data=raw_mail_data.where(pd.notnull(raw_mail_data), ' ')
```

```python
mail_data.head()
```

```python
mail_data.shape
```

### Label Encoding
#### spam ----> 0
#### ham  -----> 1

```python
mail_data.loc[mail_data['Category']=='spam', 'Category',]=0 #'Category' tells which column to update
```

```python
mail_data.loc[mail_data['Category']=='ham','Category',]=1
```

```python
mail_data.head()
```

### Data and Label

```python
X=mail_data['Message']
Y=mail_data['Category']
```

```python
print(X)
```

```python
print(Y)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,Y,test_size=0.3,random_state=3)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
# Converting Text data to Feature Vectors(Numerical) that can be used as input for Logistic Regression
```

### Feature Extraction

```python
feature_extraction=TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)
```

```python
X_train_feature=feature_extraction.fit_transform(X_train)
```

```python
X_test_feature=feature_extraction.transform(X_test)
```

#### Converting y_train and y_test to integers

```python
y_train=y_train.astype('int')
y_test=y_test.astype('int')
```

```python
print(X_train_feature)
```

### Training Model

#### Logistic Regression

```python
model=LogisticRegression()
```

```python
model.fit(X_train_feature,y_train)
```

### Evaluation

#### Accuracy on Training Data

```python
X_train_prediction=model.predict(X_train_feature)
training_accuracy=accuracy_score(y_train,X_train_prediction)
print("Training Data Accuracy: ", training_accuracy)
```

```python
X_test_prediction=model.predict(X_test_feature)
testing_accuracy=accuracy_score(y_test,X_test_prediction)
print("Testing Data Accuracy: ", testing_accuracy)
```

### Building Predictive System

```python
input_mail=["Nah I don't think he goes to usf, he lives around here though"]
```

```python
input_feature=feature_extraction.transform(input_mail)
```

```python
input_prediction=model.predict(input_feature)
```

```python
if input_prediction[0]==1:
    print("This mail is Ham (Not Spam)")
else:
    print("This mail is Spam")
```
